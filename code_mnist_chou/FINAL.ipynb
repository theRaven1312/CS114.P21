```python
import os
import re
import torch
from PIL import Image
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms
import pyheif
import io
import random

class DigitDataset(Dataset):
    def __init__(self, image_paths, transform=None):
        self.image_paths = image_paths
        self.transform = transform

    def __len__(self):
        return len(self.image_paths)

    def __getitem__(self, idx):
        img_path = self.image_paths[idx]
        try:
            if img_path.lower().endswith(".heic"):
                heif_file = pyheif.read(img_path)
                image = Image.open(io.BytesIO(heif_file.data))
            else:
                image = Image.open(img_path)
        except Exception as e:
            image = Image.new("RGB", (64, 64))

        image = image.convert("RGB")

        if self.transform:
            image = self.transform(image)

        filename = os.path.basename(img_path)
        match = re.match(r"(\d+)", filename)
        label = int(match.group(1)) if match else -1

        return image, label
```

```python
train_transform = transforms.Compose([
    transforms.Resize((64, 64)),
    transforms.RandomRotation(10),
    transforms.ColorJitter(brightness=0.2, contrast=0.2),
    transforms.ToTensor(),
    transforms.Normalize([0.5]*3, [0.5]*3)
])

test_transform = transforms.Compose([
    transforms.Resize((64, 64)),
    transforms.ToTensor(),
    transforms.Normalize([0.5]*3, [0.5]*3)
])
```

```python
def show_blurry_images(image_paths, threshold=100.0, max_display=100, title="Blurry Images"):
    blurry_imgs = []
    for path in image_paths:
        img = cv2.imread(path, 0)
        if img is None:
            continue
        laplacian_var = cv2.Laplacian(img, cv2.CV_64F).var()
        if laplacian_var < threshold:
            blurry_imgs.append((img, os.path.basename(path), laplacian_var))
            if len(blurry_imgs) >= max_display:
                break

    if blurry_imgs:
        print(f"Phát hiện {len(blurry_imgs)} ảnh có thể bị mờ:")
        plt.figure(figsize=(14, 4))
        for i, (img, fname, var) in enumerate(blurry_imgs):
            plt.subplot(1, len(blurry_imgs), i + 1)
            plt.imshow(img, cmap='gray')
            plt.title(f"{fname[0]} ({var:.1f})")
            plt.axis('off')
        plt.suptitle(title, fontsize=16)
        plt.tight_layout()
        plt.show()
    else:
        print("không phát hiện ảnh mờ.")

show_blurry_images(train_paths, threshold=100.0, max_display=15, title="Train - Blurry Images")
show_blurry_images(test_paths, threshold=100.0, max_display=15, title="Test - Blurry Images")
```

Cách tính độ mờ của ảnh phổ dùng biến thiên Laplacian (Laplacian Variance): chất lượng ảnh đầu vào đóng vai trò quan trọng trong hiệu quả học và dự đoán. Một trong những yếu tố ảnh hưởng đến chất lượng này là độ sắc nét (sharpness) của ảnh – nếu ảnh bị mờ (blurred) hoặc crop sai vị trí, mô hình sẽ khó nhận diện đặc trưng, dẫn đến dự đoán sai.

Để đánh giá mức độ mờ của ảnh, ta sử dụng chỉ số phương sai của đạo hàm bậc hai Laplacian – gọi là Laplacian Variance. Phương pháp này hoạt động theo nguyên lý sau:

Khi ảnh sắc nét, các cạnh (edges) rõ ràng và nhiều chi tiết nhỏ sẽ tạo ra sự thay đổi cường độ pixel lớn, làm cho giá trị đạo hàm bậc hai của ảnh (Laplacian) có phương sai lớn.

Ngược lại, khi ảnh mờ (do mất nét, rung, out-focus...), biên và chi tiết bị mất, khiến Laplacian có ít dao động → phương sai nhỏ.

Cách tính cụ thể: Ảnh được chuyển sang grayscale.
* Áp dụng toán tử Laplacian để làm nổi bật các vùng cạnh.
* Tính phương sai (variance) của kết quả Laplacian này.
* Chỉ số này được dùng để đo độ sắc nét.

Trong thực tế:

Ảnh có Laplacian Variance < 100 thường được coi là mờ. Nếu áp dụng cho toàn bộ tập dữ liệu huấn luyện, ta có thể lọc ra các ảnh mờ để loại bỏ, làm sạch, hoặc cảnh báo trước khi huấn luyện.

# **Giả sử lọc ảnh mờ**
Tỷ lệ ảnh mờ rất cao (~63%), cho thấy dữ liệu ban đầu có chất lượng thị giác khá thấp. Điều này có thể ảnh hưởng nghiêm trọng đến hiệu suất mô hình, đặc biệt là ResNet vốn phụ thuộc vào đặc trưng cạnh và kết cấu rõ nét.

Việc lọc ảnh mờ là hợp lý để tránh mô hình học sai hoặc bị nhiễu do dữ liệu không rõ nét. Nhưng sẽ cần bổ sung dữ liệu rõ nét hơn nếu muốn đảm bảo độ chính xác cao và tránh overfitting do dữ liệu còn lại sau lọc quá ít (đặc biệt test chỉ còn 224 ảnh).

=> Không tiến hành lọc ảnh để đảm bảo đủ số lượng ảnh học được.

```python
def filter_blurry_images(image_paths, threshold=50.0):
    """Loại bỏ những ảnh mờ khỏi danh sách và in số lượng ảnh giữ lại."""
    filtered_paths = []
    removed = 0
    for path in image_paths:
        img = cv2.imread(path, 0)  # đọc ảnh xám
        if img is None:
            continue
        laplacian_var = cv2.Laplacian(img, cv2.CV_64F).var()
        if laplacian_var >= threshold:
            filtered_paths.append(path)
        else:
            removed += 1
    print(f"Giữ lại {len(filtered_paths)} / {len(image_paths)} ảnh (đã loại {removed} ảnh mờ, threshold={threshold}).")
    return filtered_paths

train_paths = filter_blurry_images(train_paths, threshold=50.0)
test_paths = filter_blurry_images(test_paths, threshold=50.0)

train_dataset = DigitDataset(train_paths, transform=train_transform)
test_dataset = DigitDataset(test_paths, transform=test_transform)

train_loader = DataLoader(train_dataset, batch_size=64, shuffle=True, num_workers=2)
test_loader = DataLoader(test_dataset, batch_size=64, shuffle=False)
```

### Build model

# **Sử dụng ResNet18**
Phương pháp được sử dụng trong đoạn là fine-tuning một mô hình ResNet18 đã được huấn luyện sẵn (pretrained) trên tập dữ liệu lớn ImageNet để áp dụng cho một bài toán phân loại ảnh mới gồm 10 lớp.

Cụ thể, ResNet18 được tải cùng trọng số pretrained nhằm tận dụng khả năng trích xuất đặc trưng đã học được từ dữ liệu trước đó.
Sau đó, lớp fully connected cuối cùng (fc) của mô hình được thay thế bằng một lớp mới tương ứng với số lớp của bài toán phân loại chữ số viết tay (10 lớp).

Mô hình sau đó được huấn luyện lại trên tập dữ liệu mới (dataset tự tạo).

```python
import torchvision.models as models
import torch.nn as nn

class ResNet18Classifier(nn.Module):
    def __init__(self, num_classes=10): #(10 lớp)
        super(ResNet18Classifier, self).__init__()
        self.model = models.resnet18(pretrained=True)
        self.model.fc = nn.Linear(self.model.fc.in_features, num_classes)

    def forward(self, x):
        return self.model(x)
```

### Training

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = ResNet18Classifier().to(device)

optimizer = torch.optim.Adam(model.parameters(), lr=1e-4, weight_decay=1e-4)
criterion = nn.CrossEntropyLoss()
```

```python
import torch
from tqdm import tqdm

num_epochs = 20
train_acc_list = []
test_acc_list = []
train_loss_list = []
test_loss_list = []

best_acc = 0.0
best_model_wts = None

for epoch in range(num_epochs):
    # --------- TRAIN ----------
    model.train()
    correct = total = 0
    running_loss = 0.0

    for images, labels in tqdm(train_loader, desc=f"Epoch {epoch+1}/{num_epochs} - Training"):
        images, labels = images.to(device), labels.to(device)
        outputs = model(images)
        loss = criterion(outputs, labels)

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

        running_loss += loss.item() * labels.size(0)

        _, predicted = torch.max(outputs, 1)
        correct += (predicted == labels).sum().item()
        total += labels.size(0)

    train_acc = correct / total
    train_loss = running_loss / total
    train_acc_list.append(train_acc)
    train_loss_list.append(train_loss)

    # --------- EVAL ----------
    model.eval()
    correct = total = 0
    running_loss = 0.0

    with torch.no_grad():
        for images, labels in tqdm(test_loader, desc=f"Epoch {epoch+1}/{num_epochs} - Testing"):
            images, labels = images.to(device), labels.to(device)
            outputs = model(images)
            loss = criterion(outputs, labels)
            running_loss += loss.item() * labels.size(0)

            _, predicted = torch.max(outputs, 1)
            correct += (predicted == labels).sum().item()
            total += labels.size(0)

    test_acc = correct / total
    test_loss = running_loss / total
    test_acc_list.append(test_acc)
    test_loss_list.append(test_loss)

    # --------- TRACK BEST MODEL ----------
    if test_acc > best_acc:
        best_acc = test_acc
        best_model_wts = model.state_dict()

    # --------- CHECKPOINT OPTION ----------
    torch.save(model.state_dict(), f'/gdrive/MyDrive/ChauNguyen12/MNIST/checkpoint_epoch{epoch+1}.pth')

    print(f"Epoch {epoch+1}/{num_epochs} - "
          f"Train Acc: {train_acc:.4f}, Loss: {train_loss:.4f} | "
          f"Test Acc: {test_acc:.4f}, Loss: {test_loss:.4f}")

# --------- SAVE BEST MODEL ----------
model_path = '/gdrive/MyDrive/ChauNguyen12/MNIST/best_model.pth'
torch.save(best_model_wts, model_path)
print(f"\n Best model saved to {model_path} with test accuracy: {best_acc:.4f}")
```

```python
# --------- PLOT ACCURACY & LOSS ---------
def plot_training_history(train_losses, test_losses, train_accuracies, test_accuracies):
    epochs = range(1, len(train_losses)+1)
    plt.figure(figsize=(12, 5))

    # Loss plot
    plt.subplot(1, 2, 1)
    plt.plot(epochs, train_losses, label='Train Loss')
    plt.plot(epochs, test_losses, label='Test Loss')
    plt.title('Loss theo Epoch')
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.legend()

    # Accuracy plot
    plt.subplot(1, 2, 2)
    plt.plot(epochs, train_accuracies, label='Train Acc')
    plt.plot(epochs, test_accuracies, label='Test Acc')
    plt.title('Accuracy theo Epoch')
    plt.xlabel('Epoch')
    plt.ylabel('Accuracy')
    plt.legend()

    plt.tight_layout()
    plt.show()

plot_training_history(train_loss_list, test_loss_list, train_acc_list, test_acc_list)
```

```python
import matplotlib.pyplot as plt

def show_wrong_predictions_from_loader(model, dataloader, device, max_display=12):
    model.eval()
    wrong_images, wrong_preds, true_labels = [], [], []

    with torch.no_grad():
        for images, labels in dataloader:
            images, labels = images.to(device), labels.to(device)
            outputs = model(images)
            _, predicted = torch.max(outputs, 1)

            for img, pred, true in zip(images, predicted, labels):
                if pred != true:
                    wrong_images.append(img.cpu())
                    wrong_preds.append(pred.item())
                    true_labels.append(true.item())
                    if len(wrong_images) >= max_display:
                        break
            if len(wrong_images) >= max_display:
                break

    if not wrong_images:
        print("Không có ảnh sai trong số ảnh kiểm tra được hiển thị.")
        return

    # Hiển thị ảnh sai
    plt.figure(figsize=(15, 5))
    for i in range(len(wrong_images)):
        img = wrong_images[i].permute(1, 2, 0).numpy()
        plt.subplot(1, len(wrong_images), i + 1)
        plt.imshow(img)
        plt.title(f"T:{true_labels[i]} / P:{wrong_preds[i]}")
        plt.axis('off')
    plt.suptitle("Ảnh bị phân loại sai", fontsize=16)
    plt.tight_layout()
    plt.show()
```

```python
import os
from torch.utils.data import Dataset
from PIL import Image

class InferenceImageDataset(Dataset):
    def __init__(self, image_paths, transform=None):
        self.transform = transform
        self.valid_samples = []

        for path in image_paths:
            try:
                with Image.open(path) as img:
                    img.verify()
                self.valid_samples.append(path)
            except Exception as e:
                print(f"Bỏ ảnh lỗi: {path} ({e})")

    def __len__(self):
        return len(self.valid_samples)

    def __getitem__(self, idx):
        path = self.valid_samples[idx]
        try:
            with Image.open(path) as image:
                image = image.convert("RGB")
        except Exception as e:
            print(f"Lỗi khi load ảnh tại __getitem__: {path} ({e})")
            image = Image.new("RGB", (64, 64), color=(255, 255, 255))  # fallback ảnh trắng

        if self.transform:
            image = self.transform(image)

        return image, path
```

```python
def is_image_file(filename):
    image_extensions = {
        '.jpg', '.jpeg', '.png', '.bmp', '.gif', '.tiff', '.tif',
        '.webp', '.heic', '.heif', '.avif',
        '.svg', '.ico',
        '.jfif', '.pjpeg', '.pjp',
        '.raw', '.cr2', '.nef', '.orf', '.sr2', '.arw', '.dng', '.rw2', '.pef',
        '.hdr', '.exr',
        '.icns',
    }
    return os.path.splitext(filename)[1].lower() in image_extensions

image_dir = "/content/10000_images"
image_paths = []

for root, _, files in os.walk(image_dir):
    for file in files:
        if is_image_file(file):
            image_paths.append(os.path.join(root, file))

print(f"Tổng số ảnh tìm được: {len(image_paths)}")
```

Điều này cho thấy mô hình bắt đầu overfit – học quá kỹ dữ liệu huấn luyện mà không khái quát tốt sang dữ liệu mới.

Để cải thiện generalization, có thể áp dụng early stopping, data augmentation mạnh hơn, dropout, và giảm số epoch huấn luyện.
